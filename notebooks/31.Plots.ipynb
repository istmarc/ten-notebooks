# Plots

```python
import ten
import numpy as np
import matplotlib.pyplot as plt
```

## Simple 2d plot

```python
x = ten.linear([100], -10., 10.)
x
```

```python
y = ten.sin(x)
y
```

```python
plt.plot(x, y, color = "blue")
plt.title("Simple 2d plot")
plt.xlabel("x")
plt.xlabel("sin(x)")
plt.show()
```
