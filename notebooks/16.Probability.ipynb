# Probability and simulation of random variables

```python
import ten
import math
import matplotlib.pyplot as plt
ten.set_seed(123)
```

## Inverse transform sampling

**Inverse cdf of an exponential distribution**

```python
l = .5
def finv(u):
    return -math.log(u) / l
```

**Sampling from an exponential distribution**

```python
n = 1000
s = ten.math.inv_sample(n, finv)
```

```python
plt.hist(s, color = "black")
plt.title("Exponential distribution")
plt.show()
```
