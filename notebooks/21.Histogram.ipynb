# Histogram

```python
import ten
import matplotlib.pyplot as plt
ten.set_seed(1234)
```

```python
x = ten.random.rand_norm([1000])
```

```python
plt.plot(x, color = "black")
plt.show()
```

## Fitting histogram

```python
ten.ml.hist(x)
```

## Low level

```python
h = ten.ml.histogram()
# Fit
h.fit(x)
# Plot
h.plot()
```

```python
# Get the hist and bins
hist,bins = h.hist()
```

```python
plt.stairs(hist.to_numpy(), bins.to_numpy(), color = "black")
plt.show()
```
