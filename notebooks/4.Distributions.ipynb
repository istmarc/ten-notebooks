# Random distributions

```python
import ten
import matplotlib.pyplot as plt
ten.set_seed(1234)
```

## Uniform

```python
unif = ten.random.uniform(0., .5)
unif.sample()
```

```python
unif.sample((2,2))
```

```python
x = unif.sample([1000])
plt.hist(x, color = "black")
plt.show()
```

## Normal

```python
norm = ten.random.normal(0., .5)
norm.sample()
```

```python
norm.sample((2,2))
```

```python
x = norm.sample([1000])
plt.hist(x, color = "black")
plt.show()
```

## Gamma

```python
gamma = ten.random.gamma(.5, .5)
gamma.sample()
```

```python
gamma.sample((2,2))
```

```python
x = gamma.sample([1000])
plt.hist(x, color = "black")
plt.show()
```
