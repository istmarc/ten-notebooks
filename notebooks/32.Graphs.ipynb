# Graphs

```python
import ten
import matplotlib.pyplot as plt
import networkx as nx
```

## Adjacency list

```python
g = ten.graph()
```

```python
for i in range(10):
    g.add_vertex(i)
```

```python
g.add_edge(0, 1)
g.add_edge(1, 2)
g.add_edge(2, 3)
g.add_edge(0, 4)
g.add_edge(4, 5)
g.add_edge(5, 6)
g.add_edge(4, 7)
g.add_edge(0, 8)
g.add_edge(8, 9)
```

**Depth-first search**

```python
def print_vertex(v):
    print("Vertex : ", v)
```

```python
g.dfs(0, print_vertex)
```

```python
g.bfs(0, print_vertex)
```

**Adjacency matrix**

```python
m = g.to_matrix(ten.dtype.int32)
```

```python
fig, ax = plt.subplots(figsize=(3, 3))
ax.matshow(m.to_numpy())
ax.set_title("Adjacency matrix")
plt.show()
```

**Networkx**

```python
gnx = nx.graph.Graph()
vertices = 10
for i in range(vertices):
    gnx.add_node(i)
for i in range(vertices):
    for j in range(vertices):
        if i != j and m[i,j]>0:
            gnx.add_edge(i, j)
```

```python
fig, axes = plt.subplots(figsize = (5, 3))
nx.draw_planar(gnx, node_color = "grey")
pos = nx.planar_layout(gnx)
nx.draw_networkx_labels(gnx, pos)
plt.show()
```
