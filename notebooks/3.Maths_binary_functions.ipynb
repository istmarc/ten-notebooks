## Maths binary functions and operators overloading

```python
import ten
```

## Add

```python
x = ten.arange((3, 3))
y = ten.arange((3, 3))
z = x + y
z
```

## Sub

```python
x = ten.arange((3, 3), 1.)
y = ten.arange((3, 3))
z = x - y
z
```

## Div

```python
x = ten.arange((3, 3), 1.)
y = ten.arange((3, 3), 1.)
z = x / y
z
```

## Elementwise multiplication

```python
x = ten.arange([10])
y = ten.arange([10])
z = x * y
z
```

## Matrix multiplication

```python
x = ten.arange((3, 3))
y = ten.arange((3, 3))
z = x @ y
z
```

## Matrix vector multiplication

```python
x = ten.arange((3, 3))
v = ten.arange([3])
z = x @ v
z
```
