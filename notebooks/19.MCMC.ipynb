# Markov Chain Monte Carlo (MCMC)

```python
import ten
import math
import random
import matplotlib.pyplot as plt
ten.set_seed(123)
```

## MCMC for a gaussian $X \sim N(.5, 1.)$

```python
mu = .5
sigma = 1.
```

```python
# Probability density function of X
def f(x):
    v = x-mu;
    return math.exp((-1. / (2 * sigma * sigma)) * (v*v))
```

```python
# New proposed sample
def g(y):
    return y + random.uniform(-1., 1.)
```

```python
x = ten.mcmc(0., 1000, f, g)
```

```python
fig, axes = plt.subplots(figsize = (4, 2))
plt.hist(x, color = "black")
plt.show()
```

```python
# Sampling without specifying g, by default g(y) = y + unif(-1., 1.)
y = ten.mcmc(0., 1000, f)
```

```python
fig, axes = plt.subplots(figsize = (4, 2))
plt.hist(y, color = "black")
plt.show()
```
