# Combinatorics

```python
import ten
import numpy as np
import matplotlib.pyplot as plt
import scienceplots
plt.style.use("science")
```

## Factorial

```python
for i in range(10):
    print(ten.cmb.factorial(i))
```

## Permutations

```python
n = 5
P = np.zeros((n, n), np.uint64)
```

```python
for i in range(n):
    for j in range(n):
        P[i,j] = ten.cmb.perm(i,j)
```

```python
fig, axes = plt.subplots(figsize=(3,3))
axes.matshow(P)
for i in range(n):
    for j in range(n):
        axes.text(j, i, P[i,j])
plt.show()
```

## Combinations

```python
n = 5
C = np.zeros((n, n), np.uint64)
```

```python
for i in range(n):
    for j in range(i+1):
        C[i,j] = ten.cmb.comb(i,j)
```

```python
fig, axes = plt.subplots(figsize=(3,3))
axes.matshow(C)
for i in range(n):
    for j in range(n):
        axes.text(j, i, C[i,j])
plt.show()
```

## Combinations mod 2

```python
n = 100
C = ten.cmb.pascal_triangle(n)
for i in range(n+1):
    for j in range(i):
        C[i,j] = C[i,j] % 2
```

```python
Cnp = C.to_numpy()
```

```python
fig, axes = plt.subplots(figsize=(4,4))
axes.matshow(Cnp)
axes.set_title("Combinations mod 2")
plt.show()
```
