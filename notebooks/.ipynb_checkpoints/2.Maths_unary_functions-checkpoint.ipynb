# Maths with unary functions

```python
import ten
ten.set_seed(123)
```

```python
x = ten.random.rand_norm((4,4))
x
```

## Minimum and maximum

```python
ten.min(x)
```

```python
ten.max(x)
```

## Sum

```python
ten.sum(x)
```

## Product

```python
ten.prod(x)
```

## Cumulative sum

```python
ten.cum_sum(x)
```

## Absolute value

```python
ten.abs(x)
```

## Square root

```python
y = ten.abs(x)
ten.sqrt(y)
```

## Square

```python
ten.sqr(x)
```

## Sin, cos and tan

```python
ten.sin(x)
```

```python
ten.cos(x)
```

```python
ten.tan(x)
```

## Sinh, cosh and tanh

```python
ten.sinh(x)
```

```python
ten.cosh(x)
```

```python
ten.tanh(x)
```

## Asin, acos and atan

```python
y = ten.random.rand_unif((4,4), -1., 1.)
y
```

```python
ten.asin(y)
```

```python
ten.acos(y)
```

```python
ten.atan(y)
```

## Exponential

```python
ten.exp(x)
```

## Logarithm

```python
y = ten.abs(x)
```

```python
ten.log(y)
```

```python
ten.log10(y)
```

## Floor and ceil

```python
ten.floor(x)
```

```python
ten.ceil(x)
```

## Power

```python
ten.pow(x, 2)
```
