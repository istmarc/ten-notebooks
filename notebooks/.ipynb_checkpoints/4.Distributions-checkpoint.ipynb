# Random distributions

```python
import ten
```

```python
ten.set_seed(1234)
```

## Uniform

```python
unif = ten.random.uniform(0., .5)
unif.sample()
```

```python
unif.sample((2,2))
```

## Normal

```python
norm = ten.random.normal(0., .5)
norm.sample()
```

```python
norm.sample((2,2))
```

## Gamma

```python
gamma = ten.random.gamma(.5, .5)
gamma.sample()
```

```python
gamma.sample((2,2))
```
