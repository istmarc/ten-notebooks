# Ordinary Differential Equations

```python
import ten
import matplotlib.pyplot as plt
import numpy as np
import math
```

## Euler's method

```python
t = ten.linear((10), -2., 2.)
y = ten.exp(t)
```

```python
def df(t,y):
    return y
```

```python
n = 4
ty = [ten.diffeq.euler(-2., math.exp(-2.0), df, h, n) for h in [.75, .5, .25, .125, .0625]] 
```

```python
fig, axes = plt.subplots(figsize=(5,3))
plt.plot(t, y, color = "black", label = "y")
for ts,ys in ty:
    plt.plot(ts,ys)
plt.title("Euler's method")
plt.legend()
plt.show()
```
