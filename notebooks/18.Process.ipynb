# Stochastic process

```python
import ten
ten.set_seed(123)
import matplotlib.pyplot as plt
import scienceplots
plt.style.use("science")
import numpy as np
```

## Symmetric random walk

```python
n = 1000
s = ten.process.random_walk(n)
```

```python
fig, axes = plt.subplots(figsize = (5, 2))
plt.plot(s, color = "black")
plt.title("Symmetric random walk")
plt.show()
```

```python
t = 100
S = ten.process.random_walk_paths(n, t)
Snp = S.to_numpy()
```

```python
fig, axes = plt.subplots(figsize = (5, 2))
for i in range(t):
    plt.plot(Snp[:, i], alpha = .5)
plt.title("Paths of a symmetric random walk")
plt.show()
```

```python
fig, axes = plt.subplots(5, 5, figsize = (8, 4))
k = 0
for i in range(5):
    for j in range(5):
        axes[i,j].hist(Snp[k, :], density = True, color = "black")
        k += n // 25
plt.show()
```

## Brownian motion

```python
ts, w = ten.process.brownian_motion(s, n)
```

```python
fig, axes = plt.subplots(figsize = (5, 2))
plt.plot(ts, w, color = "black")
plt.title("Brownian motion")
plt.show()
```

## Brownian motion paths

```python
ts, W = ten.process.brownian_motion_paths(n, t)
```

```python
fig, axes = plt.subplots(figsize = (5, 2))
Wnp = W.to_numpy()
for i in range(n):
    plt.plot(ts, Wnp[i, :], alpha = .5)
plt.title("Brownian motion paths")
plt.show()
```

## Central limit theorem for a brownian motion

```python
E = np.zeros(t)
V = np.zeros(t)
for i in range(t):
    E[i] = Wnp[:, i].mean()
    V[i] = Wnp[:,  i].var()
```

```python
fig, axes = plt.subplots(figsize = (4, 2))
plt.plot(E, color = "black")
plt.title(r"Expectation $E[W_t]$")
plt.show()
```

```python
fig, axes = plt.subplots(figsize = (4, 2))
plt.plot(V, color = "black")
plt.title(r"Varian $V[W_t]$")
plt.show()
```

```python
fig, axes = plt.subplots(figsize = (4, 2))
plt.hist(Wnp[:, t-1], density = True, color = "black")
plt.title("Density of $W^{n}_{t}$")
plt.show()
```
