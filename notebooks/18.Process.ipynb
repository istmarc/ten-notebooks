# Stochastic process

```python
import ten
ten.set_seed(123)
import matplotlib.pyplot as plt
import numpy as np
import statsmodels.api as sm
```

## Symmetric random walk

```python
n = 10000
s = ten.process.random_walk(n)
```

```python
fig, axes = plt.subplots(figsize = (6, 3))
plt.plot(s, color = "black")
plt.title("Symmetric random walk")
plt.show()
```

```python
t = 1000
S = [None]*t
for i in range(t):
    S[i] = ten.process.random_walk(n)
```

```python
fig, axes = plt.subplots(figsize = (6, 3))
for i in range(t):
    plt.plot(S[i], alpha = .5)
plt.title("Paths of a symmetric random walk")
plt.show()
```

```python
C = [None]*n
for i in range(n):
    C[i] = np.zeros(t)
for i in range(t):
    for j in range(n):
        C[j][i] = S[i][j]
```

```python
fig, axes = plt.subplots(5, 5, figsize = (8, 4))
k = 0
for i in range(5):
    for j in range(5):
        axes[i,j].hist(C[k], density = True, color = "black")
        k += 400
plt.show()
```
