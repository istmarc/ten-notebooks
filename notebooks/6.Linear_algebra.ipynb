# Linear algebra

```python
import ten
ten.set_seed(123)
```

## Factorization routines / matrix decomposition methods

**QR factorization**

```python
x = ten.random.rand_norm((3,3))
x
```

```python
q,r = ten.linalg.qr(x)
```

```python
q
```

```python
r
```

**LU factorization**

```python
x = ten.random.rand_norm((3,3))
x
```

```python
p, l,u = ten.linalg.lu(x)
```

```python
p
```

```python
l
```

```python
u
```

**Cholesky factorization**

```python
x = ten.random.rand_norm((3,3))
x
```

```python
l,u = ten.linalg.cholesky(x)
```

```python
l
```

```python
u
```

**SVD decomposition**

```python
x = ten.random.rand_norm((3,3))
x
```

```python
u,s,vt = ten.linalg.svd(x)
```

```python
u
```

```python
s
```

```python
vt
```
