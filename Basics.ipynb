## Basic of the ten library

The ten library available at [github.com/istmarc/ten](https://github.com/istmarc/ten) offers high perfomance tensor as multidimensional arrays objects.
This is based on several numpy tutorials.

## Install the library

Follow the install instructions at [github.com/istmarc/ten](github.com/istmarc/ten).

## Basics

```python
import ten
```

```python
# Create an uninitialized vector
a = ten.tensor((3))
a
```

```python
# Or using vector
a = ten.vector(3)
a
```

```python
# Print a vector
print(a)
```

```python
# Set values
for i in range(3):
    a[i] = i
```

```python
# Access indices
for i in range(3):
    print(a[i])
```

```python
# Create a matrix or rank 2 tensor / array
b = ten.tensor((2, 3))
b
```

```python
# Or using matrix
b = ten.matrix(2, 3)
b
```

```python
# Set values
k = 0
for i in range(2):
    for j in range(3):
        b[i,j] = k
        k += 1
```

```python
b
```

```python
# FIXME Access values, right now print None
for i in range(2):
    for j in range(3):
        print(b[i, j])
```

```python
# Acces values by linear indices
# Here they are by default stored as a contiguous column
# they can also optionally be stored as a contiguous row like numpy
for i in range(2*3):
    print(b[i])
```

## Factory functions

```python
a = ten.zeros((2, 2))
a
```

```python
b = ten.ones((2,2))
b
```

```python
c = ten.fill((2,2), .5)
c
```

```python
d = ten.arange((2,2), 1.)
d
```

```python
e = ten.linear((2,2), 0., 1.)
e
```

## Data types

## Row major tensors

By default tensors / arrays are column major, but one can create row major tensors / arrays as well.

```python
a = ten.tensor.row_major((2,2))
a
```

```python
b = ten.tensor.row_major((2,2), ten.dtype.float64)
b
```
